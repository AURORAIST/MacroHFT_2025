```python
import matplotlib.pyplot as plt
import numpy as np

# -------------------------- 1. 数据准备（与你的输入完全对应）--------------------------
scenarios = ['检查点验证集', '低层k=2', 'reward_scale=30000']  # 三个场景
x = np.arange(len(scenarios))  # x轴位置（场景索引）
width = 0.25  # 柱子宽度（适配3组数据并排）

# slope指标数据（对应slope1α1、slope2α4、slope3α0）
slope_data = {
    'slope1(α=1)': [0.20, 0.20, 0.38],
    'slope2(α=4)': [0.85, 0.85, 0.75],
    'slope3(α=0)': [5.60, 5.56, 5.56]
}

# vol指标数据（对应vol1α4、vol2α1、vol3α1）
vol_data = {
    'vol1(α=4)': [0.17, 0.17, 0.49],
    'vol2(α=1)': [2.31, 2.33, 2.20],
    'vol3(α=1)': [15.17, 15.17, 12.46]
}

# -------------------------- 2. 样式配置（支持自定义调整）--------------------------
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS']  # 解决中文显示问题
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题

# 颜色配置（slope和vol对应指标颜色呼应）
slope_colors = ['#2E86AB', '#A23B72', '#F18F01']  # 深蓝、深红、橙色
vol_colors = ['#7FB069', '#FFE66D', '#FF6B6B']    # 浅绿、浅黄、浅红

# -------------------------- 3. 绘制双面板图表 --------------------------
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))  # 1行2列，画布大小16x6

# -------------------------- 左面板：slope指标 --------------------------
# 绘制3组slope柱子（并排布局）
bars1 = ax1.bar(x - width, slope_data['slope1(α=1)'], width, label='slope1(α=1)', color=slope_colors[0], alpha=0.8)
bars2 = ax1.bar(x, slope_data['slope2(α=4)'], width, label='slope2(α=4)', color=slope_colors[1], alpha=0.8)
bars3 = ax1.bar(x + width, slope_data['slope3(α=0)'], width, label='slope3(α=0)', color=slope_colors[2], alpha=0.8)

# 左面板样式调整
ax1.set_title('不同场景下的slope指标对比', fontsize=14, fontweight='bold', pad=20)
ax1.set_xlabel('场景', fontsize=12)
ax1.set_ylabel('slope数值（%）', fontsize=12)
ax1.set_xticks(x)
ax1.set_xticklabels(scenarios, rotation=0, fontsize=10)
ax1.set_ylim(0, 6)  # 适配最大5.6%，留少量余量
ax1.legend(loc='upper right', fontsize=10)
ax1.grid(axis='y', alpha=0.3, linestyle='--')

# 左面板添加数值标注（保留2位小数）
for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height + 0.05,
                f'{height:.2f}%', ha='center', va='bottom', fontsize=9)

# -------------------------- 右面板：vol指标 --------------------------
# 绘制3组vol柱子（与左面板场景对齐）
bars4 = ax2.bar(x - width, vol_data['vol1(α=4)'], width, label='vol1(α=4)', color=vol_colors[0], alpha=0.8)
bars5 = ax2.bar(x, vol_data['vol2(α=1)'], width, label='vol2(α=1)', color=vol_colors[1], alpha=0.8)
bars6 = ax2.bar(x + width, vol_data['vol3(α=1)'], width, label='vol3(α=1)', color=vol_colors[2], alpha=0.8)

# 右面板样式调整
ax2.set_title('不同场景下的vol指标对比', fontsize=14, fontweight='bold', pad=20)
ax2.set_xlabel('场景', fontsize=12)
ax2.set_ylabel('vol数值（%）', fontsize=12)
ax2.set_xticks(x)
ax2.set_xticklabels(scenarios, rotation=0, fontsize=10)
ax2.set_ylim(0, 16)  # 适配最大15.17%，留少量余量
ax2.legend(loc='upper right', fontsize=10)
ax2.grid(axis='y', alpha=0.3, linestyle='--')

# 右面板添加数值标注（保留2位小数）
for bars in [bars4, bars5, bars6]:
    for bar in bars:
        height = bar.get_height()
        ax2.text(bar.get_x() + bar.get_width()/2., height + 0.2,
                f'{height:.2f}%', ha='center', va='bottom', fontsize=9)

# -------------------------- 整体布局优化 --------------------------
plt.tight_layout()  # 自动调整子图间距，避免重叠

# -------------------------- 保存/显示图片 --------------------------
plt.savefig('slope_vol_multi_scene_visualization.png', dpi=300, bbox_inches='tight')  # 保存高清图（300dpi）
plt.show()  # 显示图片
```

```python
import matplotlib.pyplot as plt
import numpy as np

# -------------------------- 1. 数据准备（聚焦三个场景对比）--------------------------
indicators = [
    'slope1(α=1)', 'slope2(α=4)', 'slope3(α=0)',
    'vol1(α=4)', 'vol2(α=1)', 'vol3(α=1)'
]  # 6个核心指标
x = np.arange(len(indicators))
width = 0.28  # 加宽柱子，提升视觉对比感（适配3个场景并排不拥挤）

# 三个场景的核心数据（保持原数据不变，聚焦对比）
scene_data = {
    '检查点验证集': [0.20, 0.85, 5.60, 0.17, 2.31, 15.17],
    '低层k=2': [0.20, 0.85, 5.56, 0.17, 2.33, 15.17],
    'reward_scale=30000': [0.38, 0.75, 5.56, 0.49, 2.20, 12.46]
}

# -------------------------- 2. 样式优化（强化场景区分）--------------------------
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False
# 更鲜明的场景颜色（易区分、高对比度）
scene_colors = ['#1f77b4', '#ff7f0e', '#2ca02c']  # 蓝、橙、绿（matplotlib经典对比色）
fig, ax = plt.subplots(figsize=(16, 9))  # 调整画布比例，更适合对比展示

# -------------------------- 3. 绘制场景对比柱状图（核心优化）--------------------------
# 三个场景并列，每个场景对应所有指标，强化跨场景对比
bars_check = ax.bar(x - width, scene_data['检查点验证集'], width, 
                    label='检查点验证集', color=scene_colors[0], alpha=0.9, edgecolor='black', linewidth=0.5)
bars_k2 = ax.bar(x, scene_data['低层k=2'], width, 
                 label='低层k=2', color=scene_colors[1], alpha=0.9, edgecolor='black', linewidth=0.5)
bars_reward = ax.bar(x + width, scene_data['reward_scale=30000'], width, 
                     label='reward_scale=30000', color=scene_colors[2], alpha=0.9, edgecolor='black', linewidth=0.5)

# -------------------------- 4. 对比导向样式优化 --------------------------
ax.set_title('检查点验证集 vs 低层k=2 vs reward_scale=30000 场景指标对比', 
             fontsize=16, fontweight='bold', pad=30)  # 标题明确对比对象
ax.set_xlabel('指标类型（含alpha值）', fontsize=14)
ax.set_ylabel('数值（%）', fontsize=14)
ax.set_xticks(x)
ax.set_xticklabels(indicators, rotation=30, ha='right', fontsize=12)  # 减少旋转角度，提升可读性
ax.set_ylim(0, 18)  # 留足顶部空间，避免高数值标注溢出

# 强化图例（突出场景对比）
ax.legend(loc='upper left', fontsize=13, framealpha=0.9, shadow=True)

# 网格线优化（方便读取对比数值）
ax.grid(axis='y', alpha=0.4, linestyle='-', linewidth=0.8)
ax.set_axisbelow(True)  # 网格线在柱子下方，不遮挡

# -------------------------- 5. 数值标注优化（对比友好）--------------------------
def add_annotations(bars, color):
    for bar in bars:
        height = bar.get_height()
        # 标注位置自适应，避免重叠，颜色与柱子呼应
        ax.text(bar.get_x() + bar.get_width()/2., height + 0.3,
                f'{height:.2f}%', ha='center', va='bottom', 
                fontsize=10, fontweight='bold', color=color)

add_annotations(bars_check, scene_colors[0])
add_annotations(bars_k2, scene_colors[1])
add_annotations(bars_reward, scene_colors[2])

# -------------------------- 6. 增加场景差异提示（可选）--------------------------
# 在vol3位置添加差异标注（最明显的差异点）
ax.annotate('vol3显著下降', xy=(5, 12.46), xytext=(5, 14),
            arrowprops=dict(arrowstyle='->', color=scene_colors[2], lw=2),
            fontsize=11, color=scene_colors[2], fontweight='bold')

ax.annotate('slope1/vol1上升', xy=(0, 0.38), xytext=(0, 1.5),
            arrowprops=dict(arrowstyle='->', color=scene_colors[2], lw=2),
            fontsize=11, color=scene_colors[2], fontweight='bold')

# -------------------------- 7. 布局与保存 --------------------------
plt.tight_layout()
plt.savefig('three_scenes_comparison.png', dpi=300, bbox_inches='tight')  # 文件名明确对比主题
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# -------------------------- 1. 数据准备（拆分场景）--------------------------
indicators = [
    'slope1(α=1)', 'slope2(α=4)', 'slope3(α=0)',
    'vol1(α=4)', 'vol2(α=1)', 'vol3(α=1)'
]  # 统一6个指标顺序
x = np.arange(len(indicators))

# 数据拆分：双场景对比组 + 单独展示组
compare_data = {
    '检查点验证集': [0.20, 0.85, 5.60, 0.17, 2.31, 15.17],
    '低层k=2': [0.20, 0.85, 5.56, 0.17, 2.33, 15.17]
}
single_data = {
    'reward_scale=30000': [0.38, 0.75, 5.56, 0.49, 2.20, 12.46]
}

# -------------------------- 2. 全局样式配置（两张图统一风格）--------------------------
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False
colors_compare = ['#1f77b4', '#ff7f0e']  # 双场景对比色（蓝、橙）
color_single = '#2ca02c'  # 单独场景颜色（绿）
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 8))  # 1行2列，统一画布大小

# -------------------------- 3. 左图：检查点验证集 vs 低层k=2 对比 --------------------------
width_compare = 0.35  # 双场景柱子宽度（适配2组并排）
bars1 = ax1.bar(x - width_compare/2, compare_data['检查点验证集'], width_compare,
                label='检查点验证集', color=colors_compare[0], alpha=0.9, edgecolor='black', linewidth=0.5)
bars2 = ax1.bar(x + width_compare/2, compare_data['低层k=2'], width_compare,
                label='低层k=2', color=colors_compare[1], alpha=0.9, edgecolor='black', linewidth=0.5)

# 左图样式优化
ax1.set_title('检查点验证集 vs 低层k=2 指标对比', fontsize=15, fontweight='bold', pad=25)
ax1.set_xlabel('指标类型（含alpha值）', fontsize=13)
ax1.set_ylabel('数值（%）', fontsize=13)
ax1.set_xticks(x)
ax1.set_xticklabels(indicators, rotation=30, ha='right', fontsize=11)
ax1.set_ylim(0, 18)  # 与右图统一y轴范围，便于跨图对比
ax1.legend(loc='upper left', fontsize=12)
ax1.grid(axis='y', alpha=0.4, linestyle='-', linewidth=0.8)
ax1.set_axisbelow(True)

# 左图数值标注（突出细微差异）
def annotate_compare(bars, color):
    for bar in bars:
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height + 0.3,
                f'{height:.2f}%', ha='center', va='bottom',
                fontsize=9, fontweight='bold', color=color)

annotate_compare(bars1, colors_compare[0])
annotate_compare(bars2, colors_compare[1])

# 标注左图细微差异（slope3和vol2）
ax1.annotate('差异0.04%', xy=(2, 5.60), xytext=(2, 7),
            arrowprops=dict(arrowstyle='->', color='red', lw=1.5),
            fontsize=10, color='red', fontweight='bold')
ax1.annotate('差异0.02%', xy=(4, 2.33), xytext=(4, 3.5),
            arrowprops=dict(arrowstyle='->', color='red', lw=1.5),
            fontsize=10, color='red', fontweight='bold')

# -------------------------- 4. 右图：reward_scale=30000 单独展示 --------------------------
width_single = 0.6  # 单场景柱子宽度（更饱满）
bars3 = ax2.bar(x, single_data['reward_scale=30000'], width_single,
                label='reward_scale=30000', color=color_single, alpha=0.9, edgecolor='black', linewidth=0.5)

# 右图样式优化
ax2.set_title('reward_scale=30000 指标单独展示', fontsize=15, fontweight='bold', pad=25)
ax2.set_xlabel('指标类型（含alpha值）', fontsize=13)
ax2.set_ylabel('数值（%）', fontsize=13)
ax2.set_xticks(x)
ax2.set_xticklabels(indicators, rotation=30, ha='right', fontsize=11)
ax2.set_ylim(0, 18)  # 与左图统一y轴，便于对比
ax2.grid(axis='y', alpha=0.4, linestyle='-', linewidth=0.8)
ax2.set_axisbelow(True)

# 右图数值标注（突出核心特征）
for bar in bars3:
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + 0.3,
            f'{height:.2f}%', ha='center', va='bottom',
            fontsize=10, fontweight='bold', color=color_single)

# 标注右图关键特征（与左图对比的核心差异）
ax2.annotate('vol3显著降低\n（较左图低~2.7%）', xy=(5, 12.46), xytext=(5, 15),
            arrowprops=dict(arrowstyle='->', color=color_single, lw=2),
            fontsize=11, color=color_single, fontweight='bold')
ax2.annotate('slope1/vol1上升\n（较左图提升明显）', xy=(0, 0.38), xytext=(0, 2),
            arrowprops=dict(arrowstyle='->', color=color_single, lw=2),
            fontsize=11, color=color_single, fontweight='bold')

# -------------------------- 5. 全局布局与保存 --------------------------
plt.tight_layout()  # 自动调整两张图间距，避免重叠
plt.savefig('two_plots_comparison.png', dpi=300, bbox_inches='tight')  # 文件名明确双图结构
plt.show()
```
