```python
# 分类可视化 eval_best_model_val 下所有 init_pos_* 子目录的 .npy 文件（action每个init_pos单独画，reward等不加marker）
import os
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import font_manager

# 自动设置中文字体，防止乱码
zh_fonts = [f.name for f in font_manager.fontManager.ttflist if 'SimHei' in f.name or '黑体' in f.name or 'FangSong' in f.name or '宋体' in f.name]
if zh_fonts:
    plt.rcParams['font.family'] = zh_fonts[0]
else:
    plt.rcParams['font.sans-serif'] = ['Arial']
plt.rcParams['axes.unicode_minus'] = False

base_dir = '/home/ustc1958/lxy/1/MacroHFT/result/low_level/ETHUSDT/eval_best_model_val'
category_map = {}
for subdir in sorted(os.listdir(base_dir)):
    sub_path = os.path.join(base_dir, subdir)
    if not os.path.isdir(sub_path):
        continue
    npy_files = [f for f in os.listdir(sub_path) if f.endswith('.npy')]
    for fname in npy_files:
        fpath = os.path.join(sub_path, fname)
        arr = np.load(fpath, allow_pickle=True)
        if getattr(arr, 'dtype', None) == object:
            try:
                arr = np.concatenate(arr)
            except Exception:
                arr = np.array(list(arr), dtype=np.float64)
        arr = np.asarray(arr, dtype=np.float64)
        # 分类：action, reward, return_rate_mean, commission_fee_history, require_money, final_balance, action_distribution
        if 'action.npy' in fname:
            cat = 'action'
            ylabel = '动作值'
        elif 'reward.npy' in fname:
            cat = 'reward'
            ylabel = '奖励'
        elif 'return_rate_mean.npy' in fname:
            cat = 'return_rate_mean'
            ylabel = '平均收益率(%)'
        elif 'commission_fee_history.npy' in fname:
            cat = 'commission_fee_history'
            ylabel = '手续费'
        elif 'require_money.npy' in fname:
            cat = 'require_money'
            ylabel = '初始资金'
        elif 'final_balance.npy' in fname:
            cat = 'final_balance'
            ylabel = '最终资金'
        elif 'action_distribution.npy' in fname:
            cat = 'action_distribution'
            ylabel = '动作概率'
        else:
            cat = 'other'
            ylabel = '数值'
        if cat not in category_map:
            category_map[cat] = []
        category_map[cat].append((subdir, fname, arr, ylabel))

# 分类别合并绘图
for cat, items in category_map.items():
    if cat == 'action':
        # 每个init_pos单独画一张图
        for subdir, fname, arr, ylabel in items:
            plt.figure(figsize=(14, 5))
            x = np.arange(len(arr))
            plt.plot(x, arr, label=f'{subdir}', linewidth=1)
            plt.title(f'{subdir} 的动作序列')
            plt.xlabel('数据索引')
            plt.ylabel('动作值')
            plt.legend()
            plt.grid(True, alpha=0.3)
            plt.tight_layout()
            plt.show()
    else:
        # 聚合所有 init_pos_*，每个 epoch 求均值，只画一条线
        all_y = []
        for subdir, fname, arr, ylabel in items:
            if arr.ndim == 1:
                y = arr
            elif arr.ndim == 2:
                y = np.mean(arr, axis=1)
            else:
                y = arr.flatten()
            all_y.append(y)
        # 对齐长度（取最短长度）
        min_len = min(len(y) for y in all_y)
        all_y = [y[:min_len] for y in all_y]
        mean_y = np.mean(np.stack(all_y, axis=0), axis=0)
        x = np.arange(len(mean_y))
        plt.figure(figsize=(10, 5))
        if cat == 'return_rate_mean':
            plt.plot(x, mean_y, marker='o', label='所有初始仓位均值')
            plt.title('平均收益率随数据索引变化')
            plt.xlabel('数据索引')
            plt.ylabel('平均收益率(%)')
            plt.legend()
            plt.grid(True, alpha=0.3)
            plt.tight_layout()
            save_path = os.path.join(base_dir, 'return_rate_mean_by_epoch.png')
            plt.savefig(save_path)
            print(f"图已保存到 {save_path}")
            plt.show()
        else:
            plt.plot(x, mean_y, label='所有初始仓位均值')  # 不加marker
            plt.title(f'{items[0][3] if items else "均值"}随数据索引变化')
            plt.xlabel('数据索引')
            plt.ylabel(items[0][3] if items else '均值')
            plt.legend()
            plt.grid(True, alpha=0.3)
            plt.tight_layout()
            plt.show()
```
